```python
import numpy as np
import pandas as pd

data = {
    "Observation": [1, 2, 3, 4, 5, 6],
    "Rainfall": ["high", "high", "high", "low", "low", "low"],
    "SoilType": ["loamy", "sandy", "clay", "loamy", "sandy", "clay"],
    "Yield": ["good", "good", "good", "poor", "poor", "good"]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

features = ["Rainfall", "SoilType"]
errors = {}

for feature in features:
    print("\nFeature:", feature)
    total_error = 0

    for value in df[feature].unique():

        subset = df[df[feature] == value]
        majority_class = subset["Yield"].mode()[0]
        error = np.sum(subset["Yield"] != majority_class)

        total_error += error

        print(value, "->", majority_class, "Error =", error)

    errors[feature] = total_error

    accuracy = (len(df) - total_error) / len(df) * 100

    print("Total Error =", total_error)
    print("Accuracy =", accuracy, "%")

best_feature = min(errors, key=errors.get)
print("\nBest Feature:", best_feature)
print("Minimum Error:", errors[best_feature])
```
