```python
import numpy as np
import matplotlib.pyplot as plt

x1 = np.array([-1, -1, 1, 1], dtype=float)
x2 = np.array([-1, 1, -1, 1], dtype=float)
y = np.array([-2.2, -1.8, 1.8, 2.2], dtype=float)

X = np.column_stack((x1, x2))

X_b = np.c_[np.ones(len(X)), X]

alpha = 0.1
lambda_ = 0.5
iterations=3
w1=0
w2=0.03333
b=0

print("Initial weight (w) : ", w1)
print("Initial weight (w) : ", w2)
print("Initial bias (b) : ", b)
print("Learning rate : ", alpha)

theta = np.zeros(3)

cost_history = []

for i in range(iterations):
    y_pred = w1*x1+w2*x2+b
    error = y - y_pred

    # MSE gradient
    gradient = (2 / len(y)) * X_b.T.dot(error)

    # L1 regularization
    l1_gradient = lambda_ * np.sign(theta)

    # Do not regularize intercept
    l1_gradient[0] = 0

    # Update parameters
    theta = theta - alpha * (gradient + l1_gradient)

    # Cost
    mse = np.mean(error ** 2)
    l1_penalty = lambda_ * np.sum(np.abs(theta[1:]))

    cost = mse + l1_penalty
    cost_history.append(cost)

# -----------------------------
# Final Parameters
# -----------------------------
b = theta[0]
w1 = theta[1]
w2 = theta[2]

print("Intercept (b) =", b)
print("Coefficient w1 =", w1)
print("Coefficient w2 =", w2)

# -----------------------------
# Predictions
# -----------------------------
y_pred = X_b.dot(theta)

print("\nActual Y:")
print(y)

print("\nPredicted Y:")
print(y_pred)


print("\nLasso Regression Equation:")
print("Y =", round(b, 4), "+", round(w1, 4), "* X1 +",
      round(w2, 4), "* X2")


```

```python
import numpy as np

# Dataset
x = np.array([1,2,3,4], dtype=float)
y = np.array([2,2,4,5], dtype=float)

# Parameters
learning_rate = 0.05
lambda_ = 0.2
iterations = 3

# Initialize weight and bias
w = 0.0
b = 0.0

# Gradient Descent
for i in range(iterations):

    # Prediction
    y_pred = w * x + b

    # Error
    error = y_pred - y

    # Gradients
    dw = (2 / len(x)) * np.sum(x * error) + 2 * lambda_ * w
    db = (2 / len(x)) * np.sum(error)

    # Update
    w = w - learning_rate * dw
    b = b - learning_rate * db

# Prediction
y_pred = w * x + b

print("Weight (w) =", w)
print("Bias (b) =", b)

print("\nActual Y:")
print(y)

print("\nPredicted Y:")
print(y_pred)

print("\nRidge Regression Equation:")
print("Y =", round(b, 4), "+", round(w, 4), "* X")
```
