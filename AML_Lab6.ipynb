```python
import numpy as np
from scipy.stats import chi2_contingency

observed = np.array([
    [10, 5],
    [5, 10]
])
chi2, p_value, degrees_of_freedom, expected = chi2_contingency(
    observed,
    correction=False
)
alpha = 0.05
chi2_tabulated = 3.841

print("Observed Frequencies:")
print(observed)

print("\nExpected Frequencies:")
print(expected)

print("\nChi-square calculated =", chi2)
print("Degrees of freedom =", degrees_of_freedom)
print("P-value =", p_value)
if chi2 > chi2_tabulated:
    print("\nReject H0")
    print("Like Sports is significantly associated with Protein Powder.")
else:
    print("\nFail to Reject H0")
    print("Like Sports is NOT significantly associated with Protein Powder.")

if p_value < alpha:
    print("P-value < 0.05: Significant")
else:
    print("P-value >= 0.05: Not Significant")
```
